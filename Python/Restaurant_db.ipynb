```python
#Display the Top 10 Selling Menu Items

top_items = (
    df.groupby("item_name")
      .size()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,5))
top_items.plot(kind="bar")
plt.title("Top 10 Selling Menu Items")
plt.xlabel("Menu Item")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()
```

```python
#Display the Bottom 10 Selling Menu Items

bottom_items = (
    df.groupby("item_name")
      .size()
      .sort_values(ascending=True)
      .head(10)
)

plt.figure(figsize=(10,5))
bottom_items.plot(kind="barh")
plt.title("Bottom 10 Selling Menu Items")
plt.xlabel("Number of Orders")
plt.ylabel("Menu Item")
plt.show()
```

```python
#Revenue by Category

category_revenue = (
    df.groupby("category")["price"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))
category_revenue.plot(kind="bar")
plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.show()
```

```python
# Revenue Contribution Percentage

revenue_percentage = (
    df.groupby("category")["price"]
      .sum()
)

plt.figure(figsize=(7,7))
revenue_percentage.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.ylabel("")
plt.title("Revenue Contribution by Category")
plt.show()
```

```python
#Most Expensive & Least Expensive Menu Items

print("Most Expensive Item:")
print(df.loc[df["price"].idxmax(), ["item_name", "category", "price"]])

print("\nLeast Expensive Item:")
print(df.loc[df["price"].idxmin(), ["item_name", "category", "price"]])
```
